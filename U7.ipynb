```python
# wap to create a dictionary with the key as the first character and the value as a list of words starting with thst character
s="don't wait for your feelings to change to take the action. take the action and your feelings will change"
words=s.split()
d={}
for word in words:
    key=word[0]
    if key in d:
        d[key].append(word)
    else:
        d[key]=[word]
print(d)
```

```python
# wap to replace words from dictionary you are given a string replace its words from string
s="don't wait for your feelings to change to take the action. take the action and your feelings will change"
d={"to":"go","change":"no change","action":"lazy"}
l=[]
for i in s.split():
    if i in d:
        l.append(d[i])
    else:
        l.append(i)
x=" ".join(l)
print(x)
```

# set
#### mutable
#### group of unique values as a single entity
#### duplicate are not allowed
#### insertion order not preserved but we can sort the elements
#### indexing and slicing not allowed for set
#### heterogeneous objects allowed
#### we can represent set elements within curly brackets with comma separator
#### we can apply mathematical operations like union,insertion,difference etc. on set objects

### creation of set

```python
s={10,20,30,40,10,20,30,40}
print(s)
print(type(s))
```

```python
s={}
s=set()
print(type(s))
```

```python
s=[10,20,30,40,10,20,30,40]
x=set(s)
print(x)
s=list(x)
print(s)
```

```python
s=set(range(15))
print(s)
```

```python
s={1,2,3,4,1,2,3,4,5,6}
print(len(s))
```

### mathematical operation
##### 1)union

```python
s1={10,20,30,40}
s2={30,40,50,60}
print(s1.union(s2))
print(s1)
print(s2)
```

```python
print(s1|s2)
```

##### 2)intersection()

```python
s1={10,20,30,40}
s2={30,40,50,60}
print(s1.intersection(s2))
```

```python
print(s1&s2)
```

##### 3)difference()

```python
s1={10,20,30,40}
s2={30,40,50,60}
print(s1.difference(s2))
```

```python
print(s1-s2)
```

##### 4)symmetric_difference

```python
s1={10,20,30,40}
s2={30,40,50,60}
print(s1.symmetric_difference(s2))
```

##### 5)issubset()

```python
s1={"a","b","c"}
s2={"f","e","c","b","a"}
s3=s1.issubset(s2)
s4=s2.issubset(s1)
print(s3)
print(s4)
```

##### 6)issuperset()

```python
a=s1.issuperset(s2)
b=s2.issuperset(s1)
print(a)
print(b)
```
