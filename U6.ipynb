# Dictionary
- group of objects as key value pairs.     
roll no-name     
phone no-address     
- Duplicate keys are not allowed but value can be duplicated.   
- Heterogeneous objects allowed for both keys and values.  
- Insertion order is not preserved.    
- Indexing and slicing concepts are not applicable.

# creation of dictionary
- d={}  or  d=dict()

```python
d={}
d[10]='Amrutiya'
d[20]='Hensi'
d[30]='Urpit'
print(d)
```

```python
print(d[30])
```

```python
print(d[40])
```

```python
d[40]='Upi'
d[30]='Nency Didi'
print(d)
```

# Deleting element of dictionary
- del d[key]

```python
del d[10]
print(d)
del d[30]
print(d)
```

```python
del d[10]
```

```python
d.clear()
print(d)
```

```python
del d
```

```python
print(d)
```

```python
square={X:X*X for X in range(1,6) if X%2==0}
print(square)
```

```python
square={X:X*X for X in range(1,6)}
print(square)
```

# Important function of dictionary
1) dict()
2) len()
3) clear()
4) copy()
5) update()
6) get()
7) popitem()

```python
l=[(10,'Amrutiya'), (20,'Hensi'), (30,'Urpit')]
d=dict(l)
print(d)
```

```python
print(len(d))
```

```python
d1=d.copy()
```

```python
print(d1)
```

```python
d1={1:5,2:30,3:40}
d2={3:50,4:60,5:70}
d1.update(d2)
print(d1)
```

```python
d1={1:5,2:30,3:40}
d2={3:50,4:60,5:70}
d2.update(d1)
print(d2)
```

```python
d={10: 'Amrutiya', 20: 'Hensi', 30: 'Urpit'}
print(d[10])
print(d.get(10))
```

```python
print(d[40])
```

```python
print(d.get(40))
```

```python
print(d.get(10,'Nency Didi'))
print(d.get(40,'Mann'))
```

```python
d={10: 'Amrutiya', 20: 'Hensi', 30: 'Urpit'}
print(d.popitem())
```

```python
d={10: 'Amrutiya', 20: 'Hensi', 30: 'Urpit'}
k,v=d.popitem())
print(k)
print(v)
```
