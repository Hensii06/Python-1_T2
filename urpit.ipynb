## Tuple
#### -we cannot perform any changes in tuple.
#### -Tuple is read only version of list.
#### -Insertion order is preserved.
#### -Duplication are allowed.
#### -Heterogeneous objects are allowed.
#### -we can diffrentiate duplicate objects by using index.
#### -supports both +ve and -ve.
#### -represents tuple elements within parenthesis with comma seprator.
#### -paranthesis are optional but recommended to use.

```python
t=()
print(type(t))
```

```python
# it will be consider as int
t=(10)
print(type(t))
```

```python
# for single valued tuple
t=(10,)
print(type(t))
```

```python
t=20,10,20,30,40
print(t)
```

```python
t=tuple(range(10,20,2))
print(t)
```

## Accessing elements of tuple
1) by using index
2) by using slicing operator

```python
t=(10,20,30,40,50,60)
print(t[0])
print(t[3])
print(t[-2])
print(t[-1])
print(t[10])
```

```python
t=(10,20,30,40,50,60)
print(t[2:5])
print(t[2:100])
print(t[2:])
print(t[::2])
print(t[5:])
print(t[:5])
```

#### Mathematical operator for tuple
+- concatenation
*-repeatation

```python
t1=(10,20,30)
t2=(40,50,60)
t=t1+t2
print(t)
```

```python
t1=(10,20,30)
t=t1*3
print(t)
```
