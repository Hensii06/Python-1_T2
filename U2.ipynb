# find()
returns the index of first occurrence of the given substring if it is not available then we will get (-1).

```python
s='lLearning python is very easy'
print(s.find('p'))
print(s.find('e'))
# second occurons -> s.find(substring,begin,end)
print(s.find('e',3,25))
print(s.find('L'))
print(s.find('l'))
print(s.find('python'))
print(s.find('Python'))
print(s.find('java'))
```

## write a python program to display all the positions of a substring in a given string

```python
s='abcdksadejfasdwda'
sb=input('Enter a substring')
count=0
flag=False
position=-1
length=len(s)
while True:
    position=s.find(sb,position+1,length)
    if position==-1:
        break
    count=count+1
    print('Found at position:',position)
    flag=True
if flag==False:
    print('Not found')
print('The number of occurences =',count)
```

## write a program to replace each special symbol in a string with #

```python
import string
a='a@#$%%^&'
x=a.maketrans(string.punctuation,'#'*32)
print(a.translate(x))
```

## write a program marge the character of two string into a one string by taking a alternative characters

```python
s1=input("Enter string")
s2=input("Enter string")
output=''
i,j=0,0
while i<len(s1) or j<len(s2):
    output=output+s1[i]
    i=i+1
    output=output+s2[j]
    j=j+1
print(output)
```
