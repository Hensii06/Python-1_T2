### write a python program to count all the letters digits and special symbols from a given string
s='Py$@th%$13on'

```python
s='Py$@th%$13on'
digitchar=0
specialchar=0
digit=0
alpha=0
space=0
lower=0
upper=0
for i in s:
    if i.isalnum():
        digitchar=digitchar+1
        if i.isdigit():
            digit=digit+1
        else:
            alpha=alpha+1
            if i.isupper():
                upper=upper+1
            else:
                lower=lower+1
    elif i.isspace():
        space=space+1
    else:
        specialchar=specialchar+1
print('digitchar =',digitchar)
print('digit =',digit)
print('alpha =',alpha)
print('specialchar =',specialchar)
print('lower =',lower)
print('upper =',upper)
```

### write a program to find digit sum and avg from a given string
s='Py$@th173$54on389'

```python
s='Py$@th173$54on389'
count=0
sum=0
for i in s:
    if i.isdigit():
        count=count+1
        sum=sum+int(i)
print('sum =',sum)
print('avg =',sum/count)
```
