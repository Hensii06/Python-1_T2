# list comprehensions
syntax :

list=[expression for item in list if condition]

```python
s=[x*x for x in range(1,11)]
print(s)
```

```python
v=[x for x in s if x%2==0]
print(v)
```

```python
u=[2**x for x in range(1,6) if x%2==0]
print(u)
```

# write a program to print the largest even and largest odd number.

```python
l=[12,45,67,89,123]
evenlarge=-1000
oddlarge=-1000
for i in l:
    if i%2!=0:
        if oddlarge<i:
            oddlarge=i
    else:
        if evenlarge<i:
            evenlarge=i
print('even largest =',evenlarge)
print('odd largest =',oddlarge)
```

```python
l=[12,45,67,89,123]
lodd=[]
leven=[]
for i in l:
    if i%2==0:
        leven.append(i)
    else:
        lodd.append(i)
lodd.sort()
leven.sort()
print('even largest =',leven[-1])
print('odd largest =',lodd[-1])
```

# write a program to print the elements with frequency grater than a given value a
l=[1,1,1,1,2,2,3,3,3,3,5,5,5,5,6,7]  
k=2    
ans=[1,3,5]   

```python
l=[1,1,1,1,2,2,3,3,3,3,5,5,5,5,6,7]  
l1=[]
count=0
k=int(input('enter frequency'))
for i in l:
    count=l.count(i)
    if count>k and i not in l1:
        l1.append(i)
print(l1)
```
