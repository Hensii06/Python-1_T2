# lambda function
- function without any name,anonymous function or lambda function
- syntax : lambda n:n**

```python
s=lambda n:n*n
print(s(4))
print(s(5))
```

```python
s=lambda a,b:a+b
print(s(10,20))
```

# map
- for every element present in the given sequence,apply some functionality and generate new element with required modification
- syntax : map(function,sequence)

```python
l=[1,2,3,4,5]
def double(x):
    return 2*x
l1=list(map(double,l))
print(l1)
```

```python
l=[1,2,3,4,5]
l1=list(map(lambda x:2*x,l))
print(l1)
```

```python
l1=[1,2,3,4,5]
l2=[4,5,6,7,8]
l3=list(map(lambda x,y:x*y,l1,l2))
print(l3)
```

# reduce()
- reduce the sequence of elements into a single element by appling specified function.
- syntax : reduce(function,sequence)
- reduce() function present in functools module so we need to import the module to use it.

```python
from functools import*
```

```python
l=[10,20,30,40,50]
sum=reduce(lambda x,y:x+y,l)
print('sum =',sum)
mul=reduce(lambda x,y:x*y,l)
print('multiplication =',mul)
```

# filter()
- we can use filter() function to filter values from given sequence based on some condition.
- syntax : filter(function,sequence)
- where function argument is responsible to perform conditional check and sequence can be list or tuple or string

```python
def iseven(x):
    if x%2==0:
        return True
    else:
        return False
l=[5,10,15,20,25,30,35,40,45]
l1=list(filter(iseven,l))
print(l1)
```

```python
l=[5,10,15,20,25,30,35,40,45]
l1=list(filter(lambda x:x%2!=0,l))
print(l1)
```
